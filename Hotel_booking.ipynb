```python
x= df.drop(
    columns=[
        "is_canceled",
        "lead_time_group",
        "previous_cancel_group",
        "waiting_days_group",
        "assigned_room_type",
    "booking_changes",
    "room_type_changed",
    "has_booking_changes",
    "days_in_waiting_list"
    ]
)

y = df["is_canceled"]

print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
categorical_col= [
    'hotel',
    'arrival_date_year',
    'arrival_date_month',
    'meal',
    'country',
    'market_segment',
    'distribution_channel',
    'reserved_room_type',
    'deposit_type',
    'agent',
    'company',
    'customer_type'
]
numerical_col= [
    col for col in x_train.columns
    if col not in categorical_col
]
print("Categorical:", len(categorical_col))
print("Numerical:", len(numerical_col))
print(numerical_col)
```

```python
for col in categorical_col:
    x_train[col] = x_train[col].astype(str)
    x_test[col] = x_test[col].astype(str)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
preprocessor=ColumnTransformer(transformers=[("num",StandardScaler(),numerical_col),("cat",OneHotEncoder(handle_unknown="ignore"),categorical_col)])
```

```python
x_trained_processed=preprocessor.fit_transform(x_train)
x_test_processed=preprocessor.transform(x_test)
print(x_trained_processed.shape)
print(x_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
log_model=LogisticRegression(max_iter=1000,random_state=42)
log_model.fit(x_trained_processed,y_train)
```

```python
y_pred=log_model.predict(x_test_processed)
```

```python
y_pred[:10]
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score,roc_auc_score
acs=accuracy_score(y_test,y_pred)
cfm=confusion_matrix(y_test,y_pred)
pres=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
print("acc",acs)
print("cfm",cfm)
print("pres",pres)
print("recall",recall)
print("f1",f1)
```

```python
prob=log_model.predict_proba(x_test_processed)[:,1]
```

```python
roc_auc=roc_auc_score(y_test,prob)
print(roc_auc)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(random_state=42,max_depth=10)
dt_model.fit(x_trained_processed,y_train)
dty_pred=dt_model.predict(x_test_processed)
print(dty_pred[:10])
```

```python
acs=accuracy_score(y_test,dty_pred)
cfm=confusion_matrix(y_test,dty_pred)
pres=precision_score(y_test,dty_pred)
recall=recall_score(y_test,dty_pred)
f1=f1_score(y_test,dty_pred)
roc_auc_prob=dt_model.predict_proba(x_test_processed)[:,1]
roc_auc=roc_auc_score(y_test,roc_auc_prob)
print("acc",acs)
print("cfm",cfm)
print("pres",pres)
print("recall",recall)
print("f1",f1)
print("roc_auc",roc_auc)
```

```python
from sklearn.ensemble import RandomForestClassifier
ran_class=RandomForestClassifier(n_estimators=100,max_depth=10,random_state=42,n_jobs=-1)
ran_class.fit(x_trained_processed,y_train)
ran_pred=ran_class.predict(x_test_processed)
print(ran_pred[:10])
```

```python
from sklearn.metrics import (
    accuracy_score, confusion_matrix,
    precision_score, recall_score, f1_score,
    roc_auc_score
)

rf_acc = accuracy_score(y_test, ran_pred)
rf_cfm = confusion_matrix(y_test, ran_pred)
rf_pres = precision_score(y_test, ran_pred)
rf_recall = recall_score(y_test, ran_pred)
rf_f1 = f1_score(y_test, ran_pred)

rf_prob = ran_class.predict_proba(x_test_processed)[:, 1]
rf_auc = roc_auc_score(y_test, rf_prob)

print("Accuracy:", rf_acc)
print("Confusion Matrix:\n", rf_cfm)
print("Precision:", rf_pres)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
print("ROC-AUC:", rf_auc)
```

```python
#evaluating decision tree model
from sklearn.metrics import accuracy_score,f1_score
dt_model_predict=dt_model.predict(x_trained_processed)
acc1=accuracy_score(y_train,dt_model_predict)
print("training accuracy",acc1)
acc2=accuracy_score(y_test,dty_pred)
print("testing accuracy",acc2)
f1_sc1=f1_score(y_train,dt_model_predict)
print("training f1",f1_sc1)
f1_sc2=f1_score(y_test,dty_pred)
print("testing f1",f1_sc2)
```

```python
feature_names=preprocessor.get_feature_names_out()
imp_df=pd.DataFrame({"Features":feature_names,"Importance":dt_model.feature_importances_})
imp_df=imp_df.sort_values(by="Importance",ascending=False)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

param_grid = {
    "max_depth": [5, 10, 15, 20],
    "min_samples_split": [2, 10, 20],
    "min_samples_leaf": [1, 5, 10]
}

grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    scoring="f1",
    cv=3,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_trained_processed, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation F1:", grid_search.best_score_)
```

```python

best_dt_model = grid_search.best_estimator_

tuned_pred = best_dt_model.predict(x_test_processed)
tuned_prob = best_dt_model.predict_proba(x_test_processed)[:, 1]

print("Accuracy:", accuracy_score(y_test, tuned_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, tuned_pred))
print("Precision:", precision_score(y_test, tuned_pred))
print("Recall:", recall_score(y_test, tuned_pred))
print("F1 Score:", f1_score(y_test, tuned_pred))
print("ROC-AUC:", roc_auc_score(y_test, tuned_prob))
```

```python
tuned_train_pred = best_dt_model.predict(x_trained_processed)

print("Training Accuracy:",
      accuracy_score(y_train, tuned_train_pred))

print("Testing Accuracy:",
      accuracy_score(y_test, tuned_pred))

print("Training F1:",
      f1_score(y_train, tuned_train_pred))

print("Testing F1:",
      f1_score(y_test, tuned_pred))
```

```python

models = {
    "Logistic Regression": (log_model, x_test_processed, y_pred),
    "Original Decision Tree": (dt_model, x_test_processed, dty_pred),
    "Tuned Decision Tree": (best_dt_model, x_test_processed, tuned_pred),
    "Random Forest": (ran_class, x_test_processed, ran_pred)
}

results = []

for name, (model, x_eval, predictions) in models.items():
    probabilities = model.predict_proba(x_eval)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1 Score": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

comparison_df = pd.DataFrame(results)
print(comparison_df.round(4).to_string(index=False))
```
